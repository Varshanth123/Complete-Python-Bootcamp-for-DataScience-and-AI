# Boubble Sort
---------------------

### Time Complexcity

best case :O(n)

worst case:O(n^2)


```python
def BubbleSort(ele):
    n=len(ele)
    for i in range(n):
        swapped=False
        for j in range(n-i-1):
            if ele[j]>ele[j+1]:
                ele[j],ele[j+1]=ele[j+1],ele[j]
                swapped=True
        if not swapped:
            break
    return ele
```

```python
lst=[1,2,6,3,4,5,9,7,89,2,3,445]
print(BubbleSort(lst))
```

```python
names=['varshanth','Siddhu','Prem','Sunny']
print(BubbleSort(names))
```
