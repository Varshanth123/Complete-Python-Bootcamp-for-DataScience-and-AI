# Merge Sort

```python
# merge sort
class MergeSort: # O(NlogN)
    def sortArray(self, nums):
        self.Divide(nums,0,len(nums)-1)
        return nums
    def Divide(self,nums,left,right):
        if left<right:
            mid=(left+right)//2
            self.Divide(nums,left,mid)
            self.Divide(nums,mid+1,right)
            self.Conquer(nums,left,mid,right)

    def Conquer(self,nums,left,mid,right):
        i=left;j=mid+1
        lst=[]
        while(i<=mid and j<=right):
            if nums[i]<=nums[j]:
                lst.append(nums[i])
                i+=1
            else:
                lst.append(nums[j])
                j+=1
        while(i<=mid):
            lst.append(nums[i])
            i+=1
        while(j<=right):
            lst.append(nums[j])
            j+=1
        for num in lst:
            nums[left]=num
            left+=1

```

```python
srt=MergeSort()
lst=[3,1,5,3,6,7]
srt.sortArray(lst)
```
