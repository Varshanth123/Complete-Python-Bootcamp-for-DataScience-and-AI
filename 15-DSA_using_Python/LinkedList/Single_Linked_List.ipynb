# Single Linked List

```python
# Node Creation
class Node:
    def __init__(self,data,next=None):
        self.data=data
        self.next=next
```

```python
class LinkedList:
    def __init__(self):
        self.head=None
    
    def insert_begin(self,data): # O(1)
        newNode=Node(data,self.head)
        self.head=newNode
    
    def insert_end(self,data):# O(n)
        itr=self.head
        while(itr.next!=None):
            itr=itr.next
        itr.next=Node(data)
    
    def delete_begin(self): # O(1)
        if self.head!=None:
            self.head=self.head.next

    def delete_end(self): # O(n)
        if self.head==None:
            return
        if self.head.next==None:
            self.head=None
            return
        itr=self.head
        while(itr.next.next!=None):
            itr=itr.next
        itr.next=None
            

    def display_all(self): # O(n)
        itr=self.head
        while(itr != None):
            print(f'{itr.data}->',end='')
            itr=itr.next
        print('None')
    
    def count_ele(self): # O(n)
        itr=self.head
        count=0
        while(itr!=None):
            count+=1
            itr=itr.next
        return count
```

```python
ll=LinkedList()
ll.insert_begin(1)
ll.insert_begin(2)
ll.insert_begin(3)
ll.insert_begin(4)
ll.insert_begin(5)
ll.insert_begin(6)
ll.display_all()
print(ll.count_ele())
ll.delete_begin()
ll.delete_end()
ll.display_all()
print(ll.count_ele())
ll.insert_end(23)
print(ll.count_ele())
ll.display_all()
```
