# Double Linked List
-------------------------

-> compared to single linked list in double linked list reverse traversal becomes easy 

```python
# Node Creation
class Node:
    def __init__(self,data,prev=None,next=None):
        self.prev=prev
        self.data=data
        self.next=next
```

```python
class DobuleLinkedList():
    def __init__(self):
        self.head=None
        self.tail=None
    
    def insert_begin(self,data):
        newNode= Node(data,None,self.head)
        if self.head:
            self.head.prev=newNode
        self.head=newNode
        if not self.tail:
            self.tail=newNode

    def insert_end(self,data):
        newNode=Node(data,self.tail,None)
        if self.tail:
            self.tail.next=newNode
        self.tail=newNode
        if not self.head:
            self.head=newNode

    def delete_begin(self):
        if self.head:
            self.head=self.head.next
            if self.head:
                self.head.prev=None
            else:
                self.tail=None

    def delete_end(self):
        if self.tail:
            if self.tail.prev:
                self.tail=self.tail.prev
                self.tail.next=None
            else :
                self.tail=None
                self.head=None
        
    def display_forward(self):
        temp=self.head
        while temp:
            print(f'{temp.data}->',end='')
            temp=temp.next
        print('None')

    def display_reverse(self):
        temp=self.tail
        while temp:
            print(f'{temp.data}->',end='')
            temp=temp.prev
        print('None')
```

```python
dll=DobuleLinkedList()
dll.insert_begin(1)
dll.insert_begin(2)
dll.insert_begin(3)
dll.insert_begin(4)
dll.insert_end(111)
dll.display_forward()
```
