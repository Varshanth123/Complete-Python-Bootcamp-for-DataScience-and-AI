# Cirucular Single Linked List
------------------------------------

->The primary advantage of a circular linked list over a traditional linear linked list is that any node can act as the starting point for a complete list traversal

```python
# Node creation
class Node:
    def __init__(self,data,next=None):
        self.data=data
        self.next=next
```

```python
class CircularSingleLinkedList:
    def __init__(self):
        self.head=None
        self.tail=None

    def insert_begin(self,data):
        newNode=Node(data)
        if not self.head:
            self.head=newNode
            self.tail=newNode
            return
        self.tail.next=newNode
        newNode.next=self.head
        self.head=newNode

    def insert_end(self,data):
        newNode=Node(data)
        if not self.head:
            self.head=newNode
            self.tail=newNode
            return 
        self.tail.next=newNode
        newNode.next=self.head
        self.tail=newNode

    def delete_begin(self):
        if self.head:
            if self.head==self.tail:
                self.head=None
                self.tail=None
                return
            self.tail.next=self.head.next
            self.head=self.tail.next

    def delte_end(self):
        if self.head:
            if self.head==self.tail:
                self.head=None
                self.tail=None
                return 
            temp=self.head

            while temp.next!=self.tail:
                temp=temp.next
            temp.next=self.head
            self.tail=temp
    def display(self):
        if self.head:
            temp=self.head
            while True:
                print(f'{temp.data}->',end='')
                temp=temp.next
                if temp==self.head:
                    print()
                    break
```

```python
ll=CircularSingleLinkedList()
ll.insert_begin(1)
ll.insert_begin(2)
ll.display() 
ll.insert_end(0)
ll.display()
ll.delete_begin()
ll.display()
ll.delte_end()
ll.display()
```
