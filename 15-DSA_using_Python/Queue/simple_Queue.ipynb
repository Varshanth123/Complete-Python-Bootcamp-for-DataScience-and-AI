# Simple Queue
-----------------

working principal: FIFO (First in First out) 

```python
class Node:
    def __init__(self,data,next=None):
        self.data=data
        self.next=next
```

```python
class Queue:
    def __init__(self):
        self.front=None
        self.rear=None

    def enqueue(self,data):
        newNode=Node(data)
        if self.rear:
            self.rear.next=newNode
            self.rear=newNode
            return
        self.rear=newNode
        self.front=newNode

    def dequeue(self):
        if self.front:
            val=self.front.data
            if self.front==self.rear:
                self.front=None
                self.rear=None
                return val
            self.front=self.front.next
            return val
    
    def display(self):
        if self.front:
            temp=self.front
            while temp:
                print(f'{temp.data}->',end='')
                temp=temp.next
            print('None')

    def count_ele(self):
        count=0
        if self.front:
            temp=self.front
            while temp:
                count+=1
                temp=temp.next
        return count
```

```python
q=Queue()
q.enqueue(1)
q.enqueue(2)
q.enqueue(3)
q.enqueue(4)
q.enqueue(5)
q.display()
q.dequeue()
q.display()
print(q.count_ele())
```
