### Reference Counting 

Reference counting is the primary method in python uses to manage memory. Each object in python manintains a count of references pointing to it. When the refernece count drop to zero, the memory occupied by the object is deallocated.

```python
import sys # System configuration

a=[]
print(sys.getrefcount(a))
# 2 (one reference from 'a' and one from getrefcount())
```

```python
b=a
print(sys.getrefcount(a))
# 3 (1 reference from 'a', 1 reference from'b' and 1 from getrefcount())
```

```python
del b
print(sys.getrefcount(a))
```

### Garbage Collection

Python includes a cyclic collection to handle refernce cycles. Reference cycles occur when objects reference each other, preventing their reference counts from reading zero.

```python
import gc

# enable garbage collection
gc.enable()

# disable garbage collection
gc.disable()

```

```python
# The gc.collect() function in Python forces an immediate, manual garbage collection cycle to clear unreachable data from memory
gc.collect()
# the number of unreachable objects is returned
```

```python
# get garbage collection stats
print(gc.get_stats())
```

```python
# get unreachable objects
print(gc.garbage)
```

### Memory Management Best Practices

1. Use Local Variables: Local variables have a shorter lifespan and are freed sooner than global variables.
2. Avoid Circular References:Circular references can lead to memory leaks if not properly managed.
3. Use Generators: Generators produce items one at a time and only keep one item in memory at a time, making them memory efficient.
4. Explicity Delete Objects: Use the del statement to delete variables and objects explicitly.
5. Profile Memory Usage: Use memory profiling tools like tracemalloc and memory_profiler to identify memory leaks and optimize memory usage.

(note: A memory leak is a software bug that occurs when a program incorrectly manages memory, failing to release allocated space that is no longer needed.)

```python
import gc

class MyObject:
    def __init__(self, name):
        self.name = name
        print(f'Object {self.name} created')

    def __del__(self):
        print(f'Object {self.name} destroyed')

# Create circular reference
obj1 = MyObject('obj1')
obj2 = MyObject('obj2')
obj1.ref = obj2
obj2.ref = obj1

del obj1
del obj2

## Manually trigger the garbage collection
## because of the circular reference, the __del__ methods will not be called until the garbage collector runs
gc.collect()
```

```python
# Print collected objects
print(f'Number of garbage collected objects: {len(gc.garbage)}')
print(f'Garbage collected objects: {gc.garbage}')
```

### Generators for Memoery Efficiency

-> Generators allow you to produce items one at a time, using memory efficiently by only keeping one item in memory at a time.

```python
def generate_numbers(n):
    for i in range(n):
        yield i # This creates a generator that yields numbers from 0 to n-1

# using the generator
for number in generate_numbers(5):
    print(number)
    if number == 2:
        break # The generator will stop yielding numbers after 2, but it can be resumed later if needed
```

```python
# Profiling Memory usage with tracemalloc
import tracemalloc

def create_list():
    return [i for i in range(10000)]

def main():
    tracemalloc.start() # Start tracing memory allocations

    create_list() # Call the function that creates a list

    snapshot = tracemalloc.take_snapshot() # Take a snapshot of memory usage
    top_stats = snapshot.statistics('lineno') # Get statistics by line number

    print("[ Top 10 ]")
    for stat in top_stats[:10]: # Print the top 10 memory usage statistics
        print(stat)
```

```python
main()
```
