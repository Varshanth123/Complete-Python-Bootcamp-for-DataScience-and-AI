# SQL and SQLite
-------------------

SQL(Structures Query Language) is a standard language for managing and manipulatingrelational databass.

SQLite is a self-contained, serverless, and zero-configuration database engine that is widely used for embedded database systems. 
by default, in python SQLite3 comes installed

In this lession, we will cover the basics of SQL and SQLite, including creating databases, tables, and performing various SQL operations.

```python
import sqlite3
```

```python
# connect to an SQLite database
connection=sqlite3.connect('exampl.db')
connection
```

```python
cursor=connection.cursor()
```

```python
# Display
cursor.execute('''
select * from employee where id<11
''')
rows=cursor.fetchall()

for row in rows:
    print(row)
```

```python
# Update
cursor.execute('''
update employee 
    set department='Human Resource'
    where department='HR'
''')
connection.commit()

cursor.execute('''select distinct department from employee''')
dept=cursor.fetchall()
print(dept)
```

```python
# Delete 
cursor.execute('''delete from employee where id=50''')
connection.commit()

cursor.execute('''select *from employee where id>44''')
res=cursor.fetchall()
for emp in res:
    print(emp)
```

```python
# inserting multiple data 
data=[
    (50,'Varshanth',22,'AI'),
    (51,'Sunny',23,'Data Science'),
    (52,'Prudvi',21,'Data Science'),
    (53,'Mohit',26,'AI')
]

cursor.executemany('''
insert into employee(id,name,age,department)
    values(?,?,?,?)
''',data)
connection.commit()

cursor.execute('''select count(id) from employee''')
count=cursor.fetchall()
print(count)
```
