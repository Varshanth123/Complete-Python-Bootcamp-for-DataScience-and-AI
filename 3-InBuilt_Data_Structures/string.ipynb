# String handling
--------------------------
-> an immutable sequence of Unicode characters enclosed in either single, double, or triple quotation marks.


```python
a="Hello how are you"
print(a)
print(type(a))
print('length:',len(a))
```

```python
# f-strings
name="Varshanth"
age=20
print(f"My name is {name} and I am {age} years old")
```

```python
# find()->it will returns the index of en letter from left to right
print("Index of w is",a.find('w'))

#str.find(sub[, start[, end]])
#Returns the lowest index where the substring is found, or -1 if not found.
t6 = "hello world"
print(t6.find("world"))

#count()-returns number of times a character is repeated in a string 
print("'o' repeated for",a.count("o"),"times")

#str.count(sub[, start[, end]])
t2 = "hello world, helloween"
print(t2.count("ll"))
```

```python
# replace()->is used to replace a letter with given letter 
print(a.replace('o','x'))
```

```python
# lower() is used to convert all letters to lower-case
print(a.lower())

# upper() is used to convert all letters to upper-case
print(a.upper())

# capitalize()-> it will capitalize first letter 
print(a.capitalize())

#casefold()->Converts the string to lowercase. It is more aggressive than lower()
# It is particularly useful for case-insensitive comparisons.
text = "Straße"
print(text.casefold())

txt1='A'
if txt1.isupper():
    print("Yes, the character is in uppercase.")
if txt1.islower():
    print("Yes, the character is in lowercase.")
if txt1.isalpha():
    print("Yes, all the characters are alphabetic.")

if txt1.isalnum():
    print("Yes, all the characters are alphanumeric.")

if txt1.isdecimal():
    print("Yes, all the characters are decimal.")

if txt1.isdigit():
    print("Yes, all the characters are digits.")

'''isdigit() checks only digit characters, 
whereas isnumeric() is broader and also accepts other numeric characters 
such as fractions, superscripts, and numerals from various languages.

"123".isdigit()      # True
"⅔".isdigit()        # False

"123".isnumeric()    # True
"⅔".isnumeric()      # True
'''

if txt1.isidentifier():
    print("Yes, all the characters are identifiers.")

if txt1.isnumeric():
    print("Yes, all the characters are numeric.")

if txt1.isprintable():
    print("Yes, all the characters are printable.")

if txt1.isspace():
    print("Yes, all the characters are whitespace.")

if txt1.istitle():
    print("Yes, the string follows the rules of a title.")

if txt1.endswith("A"):
    print("Yes, the string ends with 'A'.")

if txt1.startswith("A"):
    print("Yes, the string starts with 'A'.")

if txt1.isascii():
    print("Yes, all the characters are ASCII.")
```

```python
#str.center(width[, fillchar])
#Centers the string within the specified width using an optional fill character.
t1 = "hello"
print(t1.center(10, '-'))

#str.encode(encoding="utf-8", errors="strict")
#Encodes the string using the specified encoding.
t3 = "hello"
encoded_text = t3.encode("utf-8")
print(encoded_text)  # Output: b'hello'

#str.endswith(suffix[, start[, end]])
#Checks if the string ends with the specified suffix.
t4 = "hello world"
print(t4.endswith("world"))  # Output: True

#str.expandtabs(tabsize=8)
#Replaces tabs in the string with spaces, using a specified tab size.
t5 = "hello\tworld"
print(t5.expandtabs(4))  # Output: "hello   world"
```

```python
# Strign SLicing is same as list slicing
a="Hello how are you"
#  0123456789....
print(a[0])
print(a[2:])
print(a[2:7])
print(a[-1])
print(a[:-2])
```

```python
# deletion of string 
del a
print(a)
```

```python
# str.split() is used to split a string into a list of substrings based on a specified delimiter.
sentence = "Hello how are you"
words = sentence.split()  # Default delimiter is whitespace 
print(words)  # Output: ['Hello', 'how', 'are', 'you']

# str.join() is used to concatenate a list of strings into a single string, using a specified separator.
separator = " "
joined_sentence = separator.join(words)
print(joined_sentence)  # Output: "Hello how are you"
```

```python
# String to list
s = "Hello how are you"
lst = list(s)
print(lst)  # Output: ['H', 'e', 'l', 'l', 'o', ' ', 'h', 'o', 'w', ' ', 'a', 'r', 'e', ' ', 'y', 'o', 'u']

# List to string
lst = ['H', 'e', 'l', 'l', 'o']
s = ''.join(lst)
print(s)  # Output: "Hello"
# with separator
s='-'.join(lst)
print(s)  # Output: "H-e-l-l-o"
```

```python
# String to tuple
s = "Hello how are you"
tup = tuple(s)
print(tup)  # Output: ('H', 'e', 'l', 'l

# tuple to string
tup = ('H', 'e', 'l', 'l', 'o')
s = ''.join(tup)
print(s)  # Output: "Hello"
```
