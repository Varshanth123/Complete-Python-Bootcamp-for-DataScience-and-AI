# Dictionary
----------------
-> key:value pair

```python
empty_dictionary=dict()
empty_dictionary
```

```python
details={
    'name':'Varshanth',
    'age':22,
    'skills':['C','Java','Python','DataBase','ML','CS core subjects'],
    'marital_satus':False,
    'Domain':'Data Science and AI'
}
details
```

```python
print(details.keys())
print(details.values())
print(details.items())
```

```python
# adding
details['On going learning']='AI/ML'
details
# if key present then it will update the value 
# if not present then it will add as key value pair 
```

```python
#pop() is used to delete an key and its respective value
details.pop('marital_satus')
details
```

```python
#update (adding an set of keys and values at a time)
details.update({'GATE_Qualified':True,'no of prb solved Leetcode':180})
details
```

```python
# copying dictionary
copied_dict=details.copy()
copied_dict
```

```python
# list.clear()->used to clear all the items 
details.clear()
details
```
