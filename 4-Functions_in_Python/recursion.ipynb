# Recursion
--------------
Recursion is a programming technique in which a function calls itself repeatedly to solve a problem by breaking it into smaller subproblems until a base case is reached.

```python
# Factorial of a number using recursion
def fact(n):
    if n==0 or n==1:
        return 1
    else:
        return n*fact(n-1)
print(fact(5))
```

```python
# fibonacci series using recursion
def fib(n):
    if n<=0:
        return 0
    elif n==1:
        return 1
    else:
        return fib(n-1)+fib(n-2)
print(fib(10))
```
