# Functions
---------------
-> a reusable block of code that executes a specific task only when called.

### Types of Arguments

1. Required arguments
2. Default arguments
3. Variable length/Arbitrary arguments
    1. *args (Variable-length positional arguments)
    2. **kwargs (Variable-length keyword arguments)
4. keyword arguments
5. Positional arguments


```python
# Sequence of arguments in a function definition should be as follows:
#  required, default, variable length, keyword only and arbitrary keyword arguments
def func(required, default=10, *args, keyword_only=None, **kwargs):
    pass
```

```python
# Required Arguments 
#-> Must be provided during the function call; otherwise, an error occurs.
def add(a,b):
    c=a+b 
    print(c)
add(12,13)
```

```python
# Positional Arguments
# -> Passed based on their position in the function call.
def display(name,age):
    print(name,age)
display("Varshanth",21)
```

```python
'''difference btw positional and required arguments
# Positional argument
->An argument passed based on its position in the function call.

# Required argument
-> A parameter that must receive a value when the function is called.
'''
# Requied arguments
def greet(name, age):
    print(name, age)

greet("John", 20)
#greet("John") we get an error

# Positional arguments
def greet(name, age):
    print(name, age)
greet("John", 20)  # Valid
# Here, "John" and 20 are positional arguments because Python assigns them based on position
greet(20, "John")  # Valid but may lead to incorrect behavior if the order is not intended
```

```python
# Default Arguments 
# -> Have a predefined value and become optional during the function call.
def display(name,course="B.tech"):
    print(name,course)
    
display(name="Varshanth")
display(name="Siddhu",course="MBBS")
```

```python
# *args (Variable-length positional arguments)
# -> Collects any extra positional arguments into a tuple.
def func(name,*sub):
    print(name,"likes to learn about")
    for s in sub:
        print(s)
func("Varshanth","c","java","python","java script","Flutter")
print()
func("Siddhu","pediatrics","General Medicen","Emergency Medicen")
```

```python
# **kwargs (Variable-length keyword arguments)
def display(**kwargs):
    print(kwargs)

display(name="John", age=20, city="Hyderabad")
```

```python
# Keyword Arguments 
# -> Can only be passed using parameter=value syntax.
def display(str,int_x,float_y):
    print("The string is = ",str)
    print("The integer is = ",int_x)
    print("The float is = ",float_y)
display(float_y=826.152,str="hello",int_x=25362)
```
