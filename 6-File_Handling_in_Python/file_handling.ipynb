### Types of all files
1. Text Files: .txt, .docx, .log etc.
2. Binary Files: .mp4, .mov, .png, .jpeg etc.

<strong>Types of Modes</strong> <br>

1. 'r' Open for reading (default)
2. 'w' Open for writing, trucating the file first (deletes the data first and writes)
3. 'x' Create a new file and open it for writing
4. 'a' Open for writing, appending to the end of the file if it exists
5. 'b' binary mode
6. 't' text mode (default)
7. '+' Open a disk file for updating (reading and writing)
8. 'r+' read + overwrite (pointer start)(no truncate)
9. 'w+' read + overwrite (pointer start)(truncate)
10. 'a+' read+append (pointer end)

```python
# if file does not exists it creates the file
f=open('demo.txt','r')
data=f.read(10) 
# 10 number of characters, if not mentioned it reads complete file
print(data)
print(type(data))
f.close()
```

```python
f=open("demo.txt",'r')
line1=f.readline()
print("First Line :",line1)
line2=f.readline()
print("Second Line : ",line2)
f.close()
```

```python
f= open('demo.txt','w')
f.write("Hey there I completed Python Programing!!!...")
f.close()
```

```python
f= open('demo.txt','r+')
print("After Write")
f.write("Wow...")
print(f.read())
f.close()
```

```python
f= open("demo.txt",'w+')
f.write("hey............")
# file content
# hey............
f.close()
```

<strong>With Syntax</strong><br>

we dont need to close the file manually it automatically closes once it comes out of with block

```python
with open("demo.txt",'a') as f:
    data=f.read()
    print(data)
```

<strong>.replace()</strong>

```python
with open('demo.txt','w') as f:
    f.write("I know java language\n")
    f.write("java is an popular language\n")
    f.write("java is OOPs language")

with open('demo.txt','r') as f:
    data=f.read()

new_data=data.replace('java','python')

with open('demo.txt','w') as f:
    f.write(new_data)

with open("demo.txt",'r') as f:
    print(f.read())
```

<strong>Searching using string .find()</strong>

```python
word='language'
with open('demo.txt','r') as f:
    data=f.read()
    if(data.find(word)!= -1):
        print('Found :)')
    else:
        print('Not Found :(')
```
