# Array Properties

```python
import numpy as np

# arr.shape returns (row,column)
arr_2d=np.array([[1,2,3],[4,5,6]])
print(arr_2d.shape)
```

```python
#arr.size returns total number of elements in the array
print(arr_2d.size)
```

```python
# arr.ndim returns the number of dimensions of the array
print(arr_2d.ndim)
```

```python
# arr.dtype returns the data type of the array
print(arr_2d.dtype)

arr=np.array([1,2.9,4,5,6])
print(arr.dtype)
# it will return float64 because it is the data type that can hold all the values in the array.
# it converts all values into highest data type in the array. In this case, it is float64 because 2.9 is a float value.
```

```python
#arr.astype() method is used to convert the data type of an array to a different data type.
arr=np.array([1.2,2.9,4.5,5.6,6.7])
int_arr=arr.astype(int)
print(int_arr)
print(int_arr.dtype)
```

# Array Operations

```python
arr=np.array([1,2,3,4,5])
print("arr + 2:", arr + 2)
print("arr - 2:", arr - 2)
print("arr * 2:", arr * 2)
print("arr / 2:", arr / 2)
print("arr ** 2:", arr ** 2)
```

```python
# Aggregation functions (summarise the data)
arr=np.array([[1,2,3],[4,5,6]])
print("Sum of all elements:", arr.sum())
print("Mean of all elements:", arr.mean())
print("max of all elements:", arr.max())
print("min of all elements:", arr.min())
print("Standard deviation of all elements:", arr.std())
print("Variance of all elements:", arr.var())
```
