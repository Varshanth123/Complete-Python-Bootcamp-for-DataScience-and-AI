# Sorting Data

sorting DAta based on 1 column <br>
df.sort_values(by="column_name",ascending=True/False,inplace=True)<br>
True -> Ascending order<br>
False -> Descending order<br>

```python
import pandas as pd

data={
    'Name':['Varshanth','John', 'Alice', 'Bob', 'Eve','Ram','Aditi','Raj','Prem','Simran'],
    'Age':[22, 30, 22, 28, 35, 27, 24, 29, 31, 26],
    'Salary':[90000, 60000, 55000, 70000, 80000, 65000, 72000, 58000, 75000, 62000],
    'Performace_Score':[96, 90, 75, 80, 95, 88, 92, 78, 83, 87]
}
df=pd.DataFrame(data)
print(df.head())
print("After sorting with respect to name")
df.sort_values(by="Salary", ascending=False, inplace=True)
print(df.head())
```

```python
df.sort_values(by=["Age","Salary"], ascending=[True,False], inplace=True)
print(df.head())
```

# Aggregation/Summary methods

1. mean() -> Average
2. median() -> Middle Value
3. mode() -> Most frequent value
4. sum() -> Sum
5. min() -> Minimum
6. max() -> Maximum
7. count() -> Conut non-null values
8. size() -> Count all values including null
9. std() -> Standard deviation
10. var() -> Vatiance
11. quantile() -> Percentile
12. nunique() -> Number of unique non-null values
13. unique() -> Return unique values
14. describe() -> Complete Statistical Summary

Note: By default, all Pandas summary methods automatically ignore null values (NaN).

```python
df['Salary'].mean()
```

```python
data={
    'Name':['Varshanth','John', 'Alice', 'Bob', 'Eve','Ram','Aditi','Raj','Prem','Simran'],
    'Age':[22, 30, 22, 28, 35, 27, 24, 29, 31, 26],
    'Salary':[90000, 60000, 55000, 70000, 80000, 65000, 72000, 58000, 75000, 62000],
    'Performace_Score':[96, 90, 75, 80, 95, 88, 92, 78, 83, 87],
    'Department':['AI','DS','FullStack','AI','DS','CyberSecurity','FullStack','Cloud','FullStack','DS']
}
df=pd.DataFrame(data)
print(df.head())
```

# Grouping 

```python
grouped=df.groupby('Department')['Salary'].mean()
print(grouped)
```
