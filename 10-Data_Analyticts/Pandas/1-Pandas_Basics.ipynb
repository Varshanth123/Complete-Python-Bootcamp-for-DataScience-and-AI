```python
data={
    "name":["varshan","sathish","karthik"],
    "age":[20,21,22],
    "city":["chennai","bangalore","hyderabad"]
}

df = pd.DataFrame(data)
print(df)

df.to_csv("output.csv",index=False)
"""
if index is true then the index will be written to the file. 
If index is false, then the index will not be written to the file.
"""
```

### Exploring the data 

head(): displays top rows (default 5 rows)<br>
tail(): displays least rows (default 5 rows)

```python
data={
    'name':["varshan","sathish","karthik",'sunny','kumar','varshanth','varun','prasad'],
    'age':[20,21,22,20,32,21,30,29],
    'city':["chennai","bangalore","hyderabad",'delhi','mumbai','pune','kolkata','chennai'],
    'salary':[20000,21000,22000,20000,32000,21000,30000,29000]
}
df=pd.DataFrame(data)
print(df.head(2)) # prints first 5 rows of the dataframe
print('---------------------------------------------------------------------------')
print(df.tail(2)) # prints last 5 rows of the dataframe
```

<strong>df.info()</strong> 

-> Gives <strong>structural information</strong> about the DataFrame.<br>
-> Shows <strong>number of rows and columns</strong> .<br>
-> Shows <strong>column names and data types</strong> .<br>
-> Shows <strong>non-null count</strong>  for each column.<br>
-> Shows <strong>memory usage</strong> .<br>

```python
df.info()
```

<strong>df.describe()</strong><br>

-> Gives <strong>statistical summary</strong> of numerical columns.<br>
-> Returns: <strong>count, mean, std, min, 25%, 50%, 75%, max.</strong><br>
-> Helps understand <strong>data distribution and outliers</strong> quickly.<br>

25% (Q1): 25% of the values are less than or equal to this value.<br>
50% (Q2 / Median): 50% of the values are less than or equal to this value.<br>
75% (Q3): 75% of the values are less than or equal to this value.

```python
df.describe() # gives the statistical summary of the dataframe
```

```python
print(df.shape) # gives the number of rows and columns in the dataframe
print()
print(df.columns) # gives the column names of the dataframe
print()
print(df.index) # gives the index of the dataframe
print()
print(df.dtypes) # gives the data types of the columns in the dataframe
print()
print(df.size) # gives the number of elements in the dataframe
print()
print(df.values) # gives the values of the dataframe as a numpy array
```

<strong>Selecting Specific column</strong>

```python
print(df['name']) # gives the selected column of the dataframe
print('---------------------------------------------------------------------------')
print(df[['name','age']]) # gives the selected columns of the dataframe
```

### Filtering rows
boolean indexing

```python
df[df['salary']>25000] # gives the rows where the salary is greater than 20000
```

& -> AND<br>
| -> OR<br>
~ -> NOT

```python
df[(df['salary']>22000) & (df['age']>25)] 
```

```python
df[df['age']>25][['name','age']] # gives the rows where the age is greater than 25 and selects the name and age columns
```
