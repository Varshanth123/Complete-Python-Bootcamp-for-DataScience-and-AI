### Modifying Data

```python
import pandas as pd
data={
    'Name':['Varshanth','John', 'Alice', 'Bob', 'Eve','Ram','Aditi','Raj','Prem','Simran'],
    'Age':[22, 30, 22, 28, 35, 27, 24, 29, 31, 26],
    'Salary':[90000, 60000, 55000, 70000, 80000, 65000, 72000, 58000, 75000, 62000],
    'Performace_Score':[96, 90, 75, 80, 95, 88, 92, 78, 83, 87]
}
df=pd.DataFrame(data)
df.head()
```

<Strong>Adding a new column</strong>

```python
df['Bonus'] = df['Salary'] * 0.1
df.head()
```

Using insert() method to add a new column at a specific position<br>
df.insert(location, "column_name", data)

```python
df.insert(0,"Employee_ID", range(1, len(df) + 1))
df.head()
```

<strong>Updating</strong><br>
.loc[]<br>
df.loc[row_index,"Column_name"] = new_value

```python
df.loc[0,'Salary']= 95000
df.head(2)
```

```python
# Increasing salary by 5%
df['Salary'] = df['Salary'] * 1.05
df.head(2)
```

<strong>Deleting Columns</strong><br>
df.drop(columns=["ColumnName"], inplace=True)

```python
df.drop(columns=['Bonus'], inplace=True)
df.head(2)
```
