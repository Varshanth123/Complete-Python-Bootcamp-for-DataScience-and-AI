# Data Visualization with Seaborn
------------------------------------
Seaborn is a Python visualization library based on Matplotlib that provides a high-level interface for drowing attractive and informative statistical graphics. Seaborn helps in creating complex visualization with just a few lines of code. 

In this lesson, we will cover the basics of Seaborn, including creating various types of plots and customizing them.

```python
import seaborn as sns
```

```python
# Basic Plotting with seaborn
tips=sns.load_dataset('tips')
tips.head()
'''
Inbuilt seaborn contains many datasets for practice
here tips is the data set which is taken from that inbuilt datasets
'''
```

```python
# create a scatter plot
import matplotlib.pyplot as plt

sns.scatterplot(x='total_bill',y='tip',data=tips)
plt.title('Scatter Plot of Toatal bill VS tips')
plt.show()
```

```python
# Line Plot

sns.lineplot(x='size',y='total_bill',data=tips)
plt.title('Line Plot of Total bill by size')
plt.show()
```

```python
# Categorical plots 
# Bar plot
sns.barplot(x='day',y='total_bill',data=tips)
plt.title('day VS MEAN(total_bill)')
plt.show()
'''
by default it will takes mean of total_bill for ploting
tips.groupby('day')['total_bill'].mean()
'''

sns.barplot(
    x='day',
    y='total_bill',
    data=tips,
    estimator=sum
)
plt.title('day VS SUM(total_bill)')
plt.show()
```

```python
# Box plot
sns.boxplot(
    x='day',
    y='total_bill',
    data=tips
)
'''
---( max value)
 |
 []
  |
---(min value)
o-> outliers
'''
```

```python
# Violin plot (it shows how the  distribution of data is)
sns.violinplot(x='day',y='total_bill',data=tips)
```

```python
# Histogram
sns.histplot(tips['total_bill'],bins=10,kde=True)
# kde smoothend cur of the distribution
```

```python
# KDE(Kernel Density Estimation) plot
sns.kdeplot(tips['total_bill'],fill=True)
```

### Pairplot  (it shows the relationship btw each and every feature)

The sns.pairplot() function in the Seaborn Python Library creates an \(N \times N\) grid of subplots to visualize pairwise relationships and distributions across all numerical columns in a dataset. By default, it plots scatter plots for off-diagonal intersections to display relationships between two variables, and histograms along the diagonal to display the univariate distribution of each singular variable.

```python
# Pairplot

sns.pairplot(tips)
```

```python
# heatmap
# corrilation=relationship btw one feature to the other feature
corrilation=tips[['total_bill','tip','size']].corr()
corrilation
# if total_bill increasing by 1 unit then tip will also increase by 0.675734 unit
```

A heatmap is a color-coded data visualization tool used to quickly identify patterns, trends, and correlation strengths within a 2D matrix.

```python
sns.heatmap(corrilation,annot=True,cmap='coolwarm')
```
