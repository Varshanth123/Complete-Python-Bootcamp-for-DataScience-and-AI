# Logging module in Python
---------------------------
The logging module is Python's built-in system for tracking events, errors, and operational information. It is the professional, production-ready upgrade to using print().

The 5 Logging Levels (Severity)
By default, Python only tracks WARNING and above. You categorize logs based on urgency:

DEBUG: Detailed diagnostic info. Used only when diagnosing problems.

INFO: Confirmation that things are working as expected.

WARNING: An indication something unexpected happened, but the software is still working (e.g., "disk space low").

ERROR: A serious problem; the software failed to perform a specific function.

CRITICAL: A severe error; the program itself may crash or stop running.

```python
# Logging module is present in python by default
import logging

## configure the basic logging settings
logging.basicConfig(level=logging.DEBUG)

## log msg
logging.debug("This is a debug message")
logging.info("This is an info message")
logging.warning('This is an warning message')
logging.error("This is an Error message")
logging.critical("This is a critical message")
```

```python
import logging
## configuring logging
logging.basicConfig(
    level=logging.DEBUG,
    format='%(asctime)s-%(name)s-%(levelname)s-%(message)s',
    datefmt='%Y-%m-%d %H:%M:%S',
    # force=True
)

## log msg
logging.debug("This is a debug message")
logging.info("This is an info message")
logging.warning('This is an warning message')
logging.error("This is an Error message")
logging.critical("This is a critical message")
```

```python
import logging
## configuring logging
logging.basicConfig(
    filename='app.log',
    filemode='w',
    level=logging.DEBUG,
    format='%(asctime)s-%(name)s-%(levelname)s-%(message)s',
    datefmt='%Y-%m-%d %H:%M:%S',
    # force=True
)

## log msg
logging.debug("This is a debug message")
logging.info("This is an info message")
logging.warning('This is an warning message')
logging.error("This is an Error message")
logging.critical("This is a critical message")
```
