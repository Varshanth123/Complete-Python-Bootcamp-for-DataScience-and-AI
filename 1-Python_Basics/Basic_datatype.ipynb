# Basic Data Types in python
--------------------------
Numeric Data type
1. int
2. float
3. complex

```python
a=22
b=7

print("a is",type(a),"type variable")

c=float(a/b)
print("c is",type(c),"type variable")

print("a/b = ",round(a/b,2))

com=3+2j
print("com is",type(com),"type variable")
print(com)
print("real part:",com.real)
print("img part:",com.imag)
```
