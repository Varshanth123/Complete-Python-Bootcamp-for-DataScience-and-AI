# Type Conversion in Python
----------------------
the process of changing a value from one data type to another.


### Implicit Type conversion 
Implicit conversion in Python happens automatically when different data types are used together in an expression.<br>

->Python converts a smaller data type to a larger one when needed.<br>
->Commonly occurs when integers and floats are combined.<br>
->Conversion happens at runtime to keep results accurate.

```python
x = 10          # Integer
y = 10.6        # Float
z = x + y     

print("x:", type(x))
print("y:", type(y))
print("z =", z)
print("z :", type(z))
```

### Explicit Type Conversion

Explicit conversion, also called type casting,is when a programmer manually changes value from one data type to another 

->Done using Python’s built-in functions like int(), float(), and str().<br>
->Gives full control over how data is interpreted or processed.<br>
->Used when automatic conversion is not suitable.

```python
s = "100"  # String
a = int(s)             
print(a)
print(type(a))
```

```python
# converting a binary string to an integer
binary_string = "1010"  # This represents the binary number for 10 in decimal  
decimal_number = int(binary_string, 2)  # Convert binary string to decimal integer
print(decimal_number)  # Output: 10
```

```python
# ASCII, Hexadecimal and Octal Conversion
c = '4'
print("ASCII of '4':", ord(c))
print("56 in Hex:", hex(56))
print("56 in Octal:", oct(56))
```

```python
# String to tuple, set and list 
s = 'Varshanth'
print("To tuple:", tuple(s))
print("To set:", set(s))
print("To list:", list(s))
```
