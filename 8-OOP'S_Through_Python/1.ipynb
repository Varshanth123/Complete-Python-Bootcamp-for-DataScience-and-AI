# OOPS in Python
-------------------

To map with real world scenarios, we started using objects in code. This is called Object Oriented Programming

```python
# Class is a blueprint for creating objects. It defines a set of attributes and methods that the created objects will have.

#Creating class
class Student:
    #Constructor method
    #All classas have a function called __init__() which is always executed when the object is being initiated.
    def __init__(self,name):
        print("Constructor is called")
        print(self)
        self.name = name

#Creating object(instances) of class
student1 = Student("Mehendhikar Varshanth")
student2 = Student("Sunny")
print(student1.name)
print(student2.name)
print(student1)
```
