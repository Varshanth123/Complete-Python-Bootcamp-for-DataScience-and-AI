To map with real world scenarios, we started using objects in code. This is called Object Oriented Programming

```python
#Creating class
class Student:
    #Constructor method
    #All classas have a function called __init__() which is always executed when the object is being initiated.
    def __init__(self,name):
        print("Constructor is called")
        print(self)
        self.name = name 

#Creating object(instances) of class
student1 = Student("Mehendhikar Varshanth")
student2 = Student("Sunny")
print(student1.name)
print(student2.name)
print(student1)
```

## Types of variables

1. Instance Variable<br>
-> Belongs to a specific <strong>object</strong>.<br>
-> Usually defined using <strong>self</strong>.<br>
-> Each object can have different values.
2. Class Variable <br>
-> Belongs to the <strong>class</strong>, shared by all objects.<br>
-> Defined directly inside the class.
3. Local Variable<br>
-> Created inside a <strong>method/function</strong>.<br>
-> <strong>Exists only while that method is executing</strong>.
4. Global Variable<br>
-> Defined <strong>outside the class/functions</strong>.<br>
-> Can be accessed <strong>throughout the program</strong>.<br>

<strong>Priority</strong> (which variable Python chooses when the same variable name exists at multiple levels)<br>

Local variable<br>
      ↓<br>
Instance variable (using self.)<br>
      ↓<br>
Class variable<br>
      ↓<br>
Global variable<br>
      ↓<br>
Built-in

### Static method

-> **Does not use self or cls.<br>
-> Belongs to the class but doesn't access instance/class data automatically.<br>
-> Can be called using the class name or object.<br>
-> Useful for utility/helper functions related to the class.

```python
class Student:
    @staticmethod
    def wellcome():
        print('Welcome to Sun Shine University')

s1=Student()
s1.wellcome()
Student.wellcome()
```

### Del Keyword

-> Used to delete object properties or object itself

```python
class Animal:
    def __init__(self,name):
        self.name=name

a1=Animal('Lizard')
print(a1.name)
del a1.name
print(a1.name)
```

## Access Modifiers

1. Public → Can be accessed directly from anywhere.
2. Protected(_) → Intended for use within the class and its subclasses.
3. Private (__) → Python changes the name internally (_Student__marks), making direct access harder.

```python
class Amount:
    # self.__acc_password can not be accessed
    def __init__(self):
        self.amount=0

    def debit(self,money):
        self.amount-=money
        self.balance()

    def credit(self,money):
        self.amount+=money
        self.balance()

    def balance(self):
        print(f'Account Holder Name : {self.name} \nAccount Number : {self._acc_num} \nBalance : {self.amount}')

class Account(Amount):
    def __init__(self,acc_num,name,acc_password):
        super().__init__()
        self.name=name # Public
        self._acc_num=acc_num # Private
        self.__acc_password=acc_password # Protected

    # Getter
    def get_password(self):
        return self.__acc_password
    def get_acc_num(self):
        return self._acc_num

    #setter
    def set_password(self,p):
        self.__acc_password=p
    def set_acc_num(self,an):
        self._acc_num=an

```

```python
ac1=Account(123,'Varshanth','@BC')
ac1.credit(1000)
ac1.balance
# ac1._acc_number can not access
# ac1.__acc_password can not access

# Indirectly accesing and modifying them 
print("Account Number: ",ac1.get_acc_num())
print("Account Password: ",ac1.get_password())

ac1.set_acc_num(222)
ac1.set_password('var')
print()
print('After Changes')
print("Account Number: ",ac1.get_acc_num())
print("Account Password: ",ac1.get_password())
```
