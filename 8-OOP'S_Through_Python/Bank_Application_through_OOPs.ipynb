# Complete Banking system
-----------------------------------

```python
from datetime import datetime 
class Amount:
    def __init__(self):
        self.balance=0
        self.transaction=[]

    def debit(self,money):
        if self.balance<=0 or self.balance-money<=0:
            print('Insuficiente Balance :(')
        else:
            self.balance-=money
            record=f'on {datetime.now()} debited amount: {money} Balance: {self.balance}'
            self.transaction.append(record)
        print("Balance Amount after Debited is",self.balance)
        print()

    def credit(self,money):
        self.balance+=money
        print("Balance Amount after Credit is",self.balance)
        print()
        record=f'on {datetime.now()} credited amount: {money} Balance: {self.balance}'
        self.transaction.append(record)

    def balance_amount(self):
        print("Balance Amount :",self.balance)
        print()

    def transaction_history(self):
        print("Your Transaction History")
        print("---------------------------------")
        for record in self.transaction:
            print(record)
        print()
```

```python
a=Amount()
a.debit(1200)
a.credit(120000)
a.balance_amount()
a.debit(300)
a.debit(5050)
a.credit(120)
a.transaction_history()
```
