# Four Pillars of OOP in Python

1. <strong>Encapsulation</strong> : Bundling data and methods together inside a class.
2. <strong>Inheritance</strong> : Child class inherits properties and methods from parent class.
3. <strong>Polymorphism</strong> : Same method/interface can behave differently for different objects.
4. <strong>Abstraction</strong> : Hiding implementation details and showing only essential functionality.
