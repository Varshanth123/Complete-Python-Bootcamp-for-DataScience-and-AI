# Exceptional Handling
---------------------------

syntax:

try:<br>
    # Contains code that may raise an exception.<br>
except:<br>
    # Handles the exception if one occurs in the try block.<br>
else:<br>
    # Executes only if no exception occurs in the try block.<br>
finally:<br>
    # Executes regardless of whether an exception occurs or not.<br>

```python
try:
    num1=int(input("Enter first number: "))
    num2=int(input("Enter second number: "))
    result=num1/num2
    print("Result: ", result)
except ZeroDivisionError:
    print("Error: Cannot divide by zero.")
except ValueError:
    print("Error: Invalid input. Please enter numeric values.")
else:
    print("Division successful.")
finally:
    print("Execution completed.")
```
